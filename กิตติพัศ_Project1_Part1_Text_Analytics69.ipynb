## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่

```python
# ----------------- Your code here -----------------
# ใบ้: อ่าน JSON lines ทีละบรรทัดด้วย generator ไม่ต้องโหลดทั้งไฟล์
# def iter_jsonl(path):
#     with open(path, encoding='utf-8') as f:
#         for line in f:
#             yield json.loads(line)
```

```python
# ----------------- Your code here -----------------
from huggingface_hub import hf_hub_download

REPO = "alpindale/two-million-bluesky-posts"
FILES = ["posts_20241127_075630.jsonl",
         "posts_20241127_122447.jsonl",
         "posts_20241127_150757.jsonl"]

paths = [hf_hub_download(repo_id=REPO, filename=f,
                         repo_type="dataset", local_dir=DATA_DIR)
         for f in FILES]
print(paths)
```

```python
#อ่านไฟล์เป็นตาราง
def iter_jsonl(path):
    with open(path, encoding="utf-8") as f:
        for line in f:
            yield json.loads(line)

posts = pd.DataFrame([r for p in paths for r in iter_jsonl(p)])

print(posts.shape)
print(posts.columns.tolist())
print("uri ซ้ำ:", posts["uri"].duplicated().sum())
print(posts.isna().mean().round(3))   # สัดส่วน missing ต่อ field
posts = posts.drop_duplicates(subset="uri").copy()
posts.head()
```

```python
#ตรวจสอบจำนวนโพสต์และเวลาเริ่มโพสต์ถึงสิ้นสุด
posts["created_at"] = pd.to_datetime(posts["created_at"], errors="coerce",
                                     utc=True, format="mixed")
print("จำนวนโพสต์:", len(posts))
print("เริ่ม:", posts["created_at"].min())
print("สิ้นสุด:", posts["created_at"].max())

posts["hour_utc"] = posts["created_at"].dt.hour
hour_counts = posts["hour_utc"].value_counts().sort_index()
print("ชั่วโมงที่โพสต์มากที่สุด (UTC):", hour_counts.idxmax())
```

```python
#ตรวจสอบโพสต์ที่วันที่เพี้ยน (เช่นปี 2008) แล้วดูว่าข้อมูลกระจุกอยู่วันไหน
print(posts["created_at"].dt.date.value_counts().sort_index().head(10))
print(posts["created_at"].dt.date.value_counts().sort_index().tail(10))
print("created_at ที่แปลงไม่ได้:", posts["created_at"].isna().sum())
```

```python
# เก็บเฉพาะโพสต์ที่เวลาสมเหตุสมผล (ปรับวันที่ตามที่เห็นจากตารางข้างบน)
mask = posts["created_at"] >= "2024-11-01"
print("ตัดออก:", (~mask).sum(), "โพสต์")

posts = posts[mask].copy()
posts["hour_utc"] = posts["created_at"].dt.hour
hour_counts = posts["hour_utc"].value_counts().sort_index()

print("จำนวนโพสต์:", len(posts))
print("เริ่ม:", posts["created_at"].min())
print("สิ้นสุด:", posts["created_at"].max())
print("ชั่วโมงที่โพสต์มากที่สุด (UTC):", hour_counts.idxmax())
```

```python
daily = posts["created_at"].dt.date.value_counts().sort_index()
print(daily)
print(posts["created_at"].quantile([0.01, 0.5, 0.99]))
```

```python
#======คำตอบข้อที่1.คนโพสต์เมื่อไร=========
posts["hour_utc"] = posts["created_at"].dt.hour
posts["hour_th"] = posts["created_at"].dt.tz_convert("Asia/Bangkok").dt.hour
hour_counts = posts["hour_utc"].value_counts().sort_index()

print("จำนวนโพสต์:", len(posts))
print("เริ่ม:", posts["created_at"].min())
print("สิ้นสุด:", posts["created_at"].max())
print("ชั่วโมงพีค (UTC):", hour_counts.idxmax())
print("ชั่วโมงพีค (เวลาไทย):", posts["hour_th"].value_counts().idxmax())
```

```python
#========คำตอบ ข้อที่ 2.ใช้ภาษาอะไร: รายงานภาษาหลักและสัดส่วน เพื่อบอกว่าถ้าทำคอนเทนต์ควรรองรับภาษาใด=======
lang_col = next((c for c in ["predicted_language", "langs", "language", "lang"]
                 if c in posts.columns), None)

if lang_col:
    lang_df = posts.copy()
    lang_df["lang"] = lang_df[lang_col].apply(
        lambda x: x[0] if isinstance(x, list) and len(x) > 0 else x)
    print("ใช้คอลัมน์:", lang_col)
else:
    from langdetect import detect, DetectorFactory
    DetectorFactory.seed = RANDOM_SEED
    def get_lang(t):
        try:
            return detect(t) if isinstance(t, str) and len(t.strip()) >= 10 else None
        except Exception:
            return None
    lang_df = posts.sample(50_000, random_state=RANDOM_SEED).copy()
    lang_df["lang"] = lang_df["text"].apply(get_lang)
    print("ตรวจภาษาเองด้วย langdetect บน sample", len(lang_df))

n_missing = lang_df["lang"].isna().sum()
print("ไม่มีค่าภาษา:", n_missing, f"({n_missing/len(lang_df):.1%})")
print((lang_df["lang"].value_counts(normalize=True).head(10) * 100).round(2))
```

```python
#========คำตอบ ข้อที่ 3.คนพูดถึงอะไร: หา hashtag หรือคำ/วลีเด่นที่น่าสนใจ และยกตัวอย่างข้อความจริงประกอบ======
from nltk.corpus import stopwords

posts["hashtags"] = posts["text"].fillna("").str.findall(r"#\w+")
tags = Counter(t.lower() for hs in posts["hashtags"] for t in hs)
print("Top hashtags:", tags.most_common(20))

sw = set(stopwords.words("english"))
en = lang_df[lang_df["lang"] == "en"]["text"].fillna("")
words = Counter(w for t in en.str.lower().str.findall(r"[a-z']{3,}")
                for w in t if w not in sw)
print("Top words:", words.most_common(30))

if tags:
    top_tag = tags.most_common(1)[0][0]
    for t in posts[posts["text"].str.lower().str.contains(top_tag, na=False, regex=False)]["text"].head(3):
        print("-", t[:200])
```
